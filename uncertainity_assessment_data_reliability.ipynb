```python
import geopandas as gpd
import numpy as np
from pathlib import Path
import pandas as pd
```

Calculate the overall accuracy of the irrigation dataset using the validation data provided by the authors

```python
# Create the Reference and Original columns in the Land Management Practice file
```

```python
# Paths
infile = Path(r"revision_analysis\lmp_error_propagation\glr_fp_ir_rf.shp")
outfile = infile.with_name("glr_fp_ir_rf.csv")

# Read shapefile
gdf = gpd.read_file(infile)

# Reference class
gdf["Ref"] = np.where(
    gdf["Name"].str.startswith("ir"),
    "IR",
    "No IR"
)

# Irrigation columns
ir_cols = ["BF_Mean_Ir"] + [f"BF_Mean__{i}" for i in range(1, 10)] + [f"BF_Mean_{i}" for i in range(11, 16)]

# Original classification
gdf["Original"] = np.where(
    gdf[ir_cols].eq("IR").any(axis=1),
    "IR",
    "No IR"
)

# Remove LMP columns
gdf = gdf.drop(columns=[c for c in gdf.columns if c.startswith("LMP_")])

# Save CSV without geometry
gdf.drop(columns="geometry").to_csv(outfile, index=False)

print(outfile)
```

Calculate the overall accuracy here

```python
file = Path(r"revision_analysis\lmp_error_propagation\glr_fp_ir_rf.csv")

df = pd.read_csv(file)

TP = ((df["Ref"] == "IR") & (df["Original"] == "IR")).sum()
TN = ((df["Ref"] == "No IR") & (df["Original"] == "No IR")).sum()
FP = ((df["Ref"] == "No IR") & (df["Original"] == "IR")).sum()
FN = ((df["Ref"] == "IR") & (df["Original"] == "No IR")).sum()

OA = (TP + TN) / (TP + TN + FP + FN)

print("TP:", TP)
print("TN:", TN)
print("FP:", FP)
print("FN:", FN)
print(f"Overall Accuracy: {OA * 100:.2f}%")
```
